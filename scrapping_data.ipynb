# Scrapping Data


```python
from google_play_scraper import app, reviews, Sort, reviews_all
import pandas as pd
```

```python
ulasan, continuation_token = reviews(
    'com.grabtaxi.passenger',
    lang = 'id',
    country = 'id',
    sort = Sort.NEWEST,
    count = 50000
)
```

# Save ke format CSV


```python
df = pd.DataFrame(ulasan)
df.to_csv('ulasan_grab.csv', index=False)
```
